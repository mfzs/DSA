# Exception Handling
- Exception is something which breaks the normal workflow
- Occured during program Execution
- Handling them give advantage that our code does not break 

```python
try:
    a=b
except:
    print("variable has not assigned")
```

```python
try:
    a=b
except NameError as ex:
    print(ex)
```

```python
try:
    result=0/0
except ZeroDivisionError as ex:
    print(ex)
```

```python
try:
    result=1/2
    a=b
    print(result)
except ZeroDivisionError as ex:
    print(ex)

except Exception as master:
    print(master)
```

```python
try:
    num=int(input("Enter the Number"))
    result=10/num
    
except ValueError as ve:
    print(ve)
except ZeroDivisionError as zd:
    print(zd)
except Exception as ex:
    print(ex)
    
else: # If no exception
    print(f"The Result: {result:.2f}")

finally: # It will run always
    print("Excecution is completed")
```

```python
try:
    file=open('myfile2.txt','r')
    content=file.read()
    for line in content:
        print(line)
        
    # print(content)

except FileNotFoundError as fnf:
    print(fnf)


finally:
    if 'file' in locals() or not file.closed():
        file.close()
        print("File close")
```
