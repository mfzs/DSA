File Operations

```python
with open('myfile2.txt',mode='r') as file:
    content=file.read()
    print(content)
```

```python
with open('myfile2.txt',mode='w') as file:
    file.write("This is my file")
    print(content)
```

```python
with open('myfile2.txt',mode='a') as file:
    file.write("\nThis is my file 2")
    print(content)
```

```python
data=b'\x00\x01'

with open('binaryfile.bin','wb') as file:
    file.write(data)
```
