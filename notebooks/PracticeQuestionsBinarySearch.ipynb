```python
## Find smallest letter greater than target
'''
letters = ['c', 'f', 'j']
target = 'k'
Output: 'c'
'''

def smallest_letter(letters, target):
    for ch in letters:
        if ch > target:
            return ch
    return letters[0] 

letters = ['c', 'f', 'j','r']
target = 'q'
smallest_letter(letters,target)
```

```python
# Find First and Last Position of Element in Sorted Array
def first_last(nums,target):
    first_occurance=-1
    second_occurance=-1
    for i in range(0,len(nums)):
        if nums[i]==target:
            first_occurance=i
            # print(first_occurance)
            break
    for j in range(first_occurance,len(nums)):
        # print(j)
        # print("coming in loop")
        if nums[j]==target:
            second_occurance=j 
            # print("coming in if")
    return [first_occurance,second_occurance]
arr=[5, 7, 7, 8, 8, 10]
target=8
first_last(arr,target)
```

```python
# Minimum in Rotated Sorted Array

def min_sort_rotated(nums):
    result=nums[0]
    # print(len(nums))
    for i in range(len(nums)-1):
        # print(nums[i])
        if nums[i]>nums[i+1]:
            result=nums[i+1]
            # print(result)
            break
    return result

arr=[4,5,6] 
min_sort_rotated(arr)
```

```python
# Search in Rotated Sorted Array
def search_sort_rotated(nums,target):
    result=-1
    # print(len(nums))
    for i in range(len(nums)-1):
        # print(nums[i])
        if nums[i]==target:
            result=i
            # print(result)
            break
    return result

arr=[4, 5, 6, 7, 0, 1, 2] 
target=1
search_sort_rotated(arr,target)
```

```python
def find_max_element(lst):
    """
    Function to find the maximum element in a list.
    :param lst: List[int] -> List of integers
    :return: int -> The maximum element in the list
    """
    max=lst[0]
    for i in range(1,len(lst)):
        if max<lst[i]:
            max=lst[i]
    return max

arr=[-1, -2, -3, -4]
find_max_element(arr)
```

```python
# Plus one
'''
Input: digits = [1, 2, 3]
Output: [1, 2, 4]
 
Input: digits = [4, 3, 2, 1]
Output: [4, 3, 2, 2]
 
Input: digits = [9, 9, 9]
Output: [1, 0, 0, 0]
'''

def plus_one(digits):
    n=len(digits)
    carry=1
    for i in range(len(digits) - 1, -1, -1):
        total=digits[i]+carry
        digits[i]=total%10
        carry=total//10
        if carry == 0:
            break
    if carry:
        digits.insert(0, carry)
        
    return digits
digits = [9, 9, 9]
plus_one(digits)
```

```python
# Intersection of two Lists

def intersection_list(arr1,arr2):
    result=[]
    for i in arr2:
        if i in arr1 and i not in result:
            result.append(i)
    return result


arr1 = [4, 9, 5]
arr2 = [9, 4, 9, 8, 4]
intersection_list(arr1,arr2)
```

```python
# Max Consecutive Ones

def max_ones(arr):
    count=0
    maximum=0
    for i in range(len(arr)):
        if arr[i]==1:
            count+=1
            maximum=max(maximum,count)
        else:
            count=0
    return maximum

arr=[1, 0, 1, 1, 0, 1, 1, 1, 1,0]
max_ones(arr)
```

```python
## Maximum Subarray Sum
def max_subarray_sum(arr):
    current_sum = arr[0]
    max_sum = arr[0]
    
    for i in range(1, len(arr)):
        current_sum = max(arr[i], current_sum + arr[i])
        max_sum = max(max_sum, current_sum)
    
    return max_sum


arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]
print(max_subarray_sum(arr))
```
