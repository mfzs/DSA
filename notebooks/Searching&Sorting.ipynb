 - All array are of same data type
 - Searching in a list/array means if an elemnet is present in your dataset or not
 - - Linear Search -- Search one by one, O(n)
   - Binary Search -- Searching using two pointer, Sorted array
   - - - Binary Search O(logn)

```python
# Linear Search
def linear_search(arr,target):
    for i in range(len(arr)):
        if arr[i]==target:
            return True
    return False
arr=[1,2,3]
target=2
print(linear_search(arr,target))
```

```python
# Binary Search
def binary_search(arr,target):
    length=len(arr)
    start=0
    mid=length//2
    end=length-1
    
    while(start<=end):
        mid=(start+end)//2
        if arr[mid]==target:
            return True
        elif arr[mid]>target:
            end=mid-1
        elif arr[mid]<target:
            start=mid+1
    return False
arr=[1,2,3,4,10,20,30,40]
target=40
print(binary_search(arr,target))
```

```python
'''
Bubble Sort
- Compare two adjacent elements
- Runs for n-1 times even for sorted array
- T.C. O(n^2)
'''

def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr


arr=[64, 34, 25, 12, 22, 11, 90]
bubble_sort(arr)
```

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped=False
        # print(i)
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped=True
        if not swapped: # Means if array got sorted in 1 or 2 runs dont run for multiple loops
            break
    return arr
arr=[64, 34, 25, 12, 22, 90, 100]
bubble_sort(arr)
```

```python
'''
Selection Sort
- Select each and every element and put them on their location
- Swap the min element with the position
- 
'''

def selection_sort(arr):
    n=len(arr)
    for i in range(n-1):
        min_index=i
        for j in range(i+1,n):
            if arr[j]<arr[min_index]:
                min_index=j
        arr[i],arr[min_index] = arr[min_index],arr[i]              
            
    return arr
arr=[64, 34, 25, 12, 22, 90, 100]
selection_sort(arr)
```

```python
'''
Insertion Sort
- Compare the next element with the previous one and put it on the right position
- 
'''

def insertion_sort(lst):
    n=len(lst) 
    for current in range(1,n):
        currentCard=lst[current]
        correctPosition=current-1

        while correctPosition>=0:
            if lst[correctPosition]<currentCard:
                break
            else:
                lst[correctPosition+1]=lst[correctPosition]
                correctPosition-=1 
            lst[correctPosition+1]=currentCard
    return lst
    
arr=[64, 34, 25, 13, 22, 90, 100]
insertion_sort(arr)
```
