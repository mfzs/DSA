```python
# Linear Search
def linear_search(arr,target):
    for i in range(len(arr)):
        if arr[i]==target:
            return True
    return False
arr=[1,2,3]
target=
print(linear_search(arr,target))
```

```python
# Binary Search
def binary_search(arr,target):
    length=len(arr)
    start=0
    mid=length//2
    end=length-1
    
    while(start<=end):
        mid=(start+end)//2
        if arr[mid]==target:
            return True
        elif arr[mid]>target:
            end=mid-1
        elif arr[mid]<target:
            start=mid+1
    return False
arr=[1,2,3,4,10,20,30,40]
target=3
print(binary_search(arr,target))
```
