```python
## merge_dicts_with_overlapping_keys
dict1={'a': 1, 'b': 2}
dict2={'b': 3, 'c': 4}
dict3={'c': 5, 'd': 6}
#output={'a': 1, 'b': 5, 'c': 9, 'd': 6}

def sum_keys(dict1,dict2,dict3):
    result={}
    for i in dict1:
        result[i]=dict1[i]
        # print(i)
    for i in dict2:
        if i in result:
            result[i]=result[i]+dict2[i]
        else:
            result[i]=dict2[i]
    for i in dict3:
        if i in result:
            result[i]=result[i]+dict3[i]
        else:
            result[i]=dict3[i]
    return result
        
print(sum_keys(dict1,dict2,dict3))
```

```python
# GCD of two numbers
def gcd(num1,num2):
    for i in range(2,min(num2,num1)+1):
        if num1%i==0 and num2%i==0:
            result=i
            # print(i)
    return result

gcd(56,56)
```
