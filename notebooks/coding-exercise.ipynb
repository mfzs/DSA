```python
# LINEAR SEARCH
def linear_search(arr,target):
    result=-1
    index=-1
    for i in range(len(arr)):
        if arr[i]==target:
            result=arr[i]
            index=i
            break
    return f"Target {result} at index {index}"
    
arr=[2,3,4,2,5,6]
# arr=[]
target=2
print(linear_search(arr,target))
```

```python
## Sum od the list
def sum(arr):
    sum =0
    for i in range(len(arr)):
        sum=sum+arr[i]
    return sum
    
arr=[2,3,4,2,5,6]
print(sum(arr))
```

```python
## Largest number
def largest_numbers(arr):
    max=arr[0]
    for i in range(len(arr)):
        if max<arr[i]:
            max=arr[i]
    return max
# arr=[2,3,4,12,5,6]
arr = [-5, -10, -2, -1, -4]
print(largest_numbers(arr))
```

```python
## remove Duplicate
def remove_duplicate(arr):
    temp=[]
    for i in range(len(arr)):
        # print(arr[i])
        if arr[i] not in temp:
            # print(arr[i])
            temp.append(arr[i])
    return temp
arr=[2,3,4,2,5,6,0,0,-1,-1]
print(remove_duplicate(arr))
```

```python
def max_consecutive_difference(lst):
    max_diff=0
    for i in range(len(lst)-1):
        tmax=abs(lst[i]-lst[i+1])
        # print(tmax)
        if tmax>max_diff:
            max_diff=tmax
    return max_diff

lst = [1, 7, 3, 10, 25]
max_consecutive_difference(lst)
```

```python
def merge_two_sorted_lists(list1, list2):
    temp=[]
    i,j=0,0
    len_list1=len(list1)
    len_list2=len(list2)

    while i<len_list1 and j<len_list2:
        if list1[i]<list2[j]:
            temp.append(list1[i])
            i=i+1
        else:
            temp.append(list2[j])
            j=j+1
    while i<len_list1:
        temp.append(list1[i])
        i+=1
    while j<len_list2:
        temp.append(list2[j])
        j+=1
        
    return temp
    
list1 = [1,3,5,7,9]
list2 = [2,4,6,8]
print(merge_two_sorted_lists(list1,list2))
```
