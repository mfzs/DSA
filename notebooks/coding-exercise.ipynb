```python
# LINEAR SEARCH
def linear_search(arr,target):
    result=-1
    index=-1
    for i in range(len(arr)):
        if arr[i]==target:
            result=arr[i]
            index=i
            break
    return f"Target {result} at index {index}"
    
arr=[2,3,4,2,5,6]
# arr=[]
target=2
print(linear_search(arr,target))
```

```python
# LINEAR SEARCH
def linear_search(arr,target):
    for i in range(len(arr)):
        if arr[i]==target:
            return f"Target {arr[i]} at index {i}"
    return "Value not found"
    
arr=[2,3,4,2,5,6]
# arr=[]
target=7
print(linear_search(arr,target))
```

```python
## Sum od the list
def sum(arr):
    sum =0
    for i in range(len(arr)):
        sum=sum+arr[i]
    return sum
    
arr=[2,3,4,2,5,6]
print(sum(arr))
```

```python
## Largest number
def largest_numbers(arr):
    max=arr[0]
    for i in range(len(arr)):
        if max<arr[i]:
            max=arr[i]
    return max
# arr=[2,3,4,12,5,6]
arr = [-5, -10, -2, -1, -4]
print(largest_numbers(arr))
```

```python
## remove Duplicate
def remove_duplicate(arr):
    temp=[]
    for i in range(len(arr)):
        # print(arr[i])
        if arr[i] not in temp:
            # print(arr[i])
            temp.append(arr[i])
    return temp
arr=[2,3,4,2,5,6,0,0,-1,-1]
print(remove_duplicate(arr))
```
