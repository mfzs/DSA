```python
# LINEAR SEARCH
def linear_search(arr,target):
    result=-1
    index=-1
    for i in range(len(arr)):
        if arr[i]==target:
            result=arr[i]
            index=i
            break
    return f"Target {result} at index {index}"
    
arr=[2,3,4,2,5,6]
# arr=[]
target=2
print(linear_search(arr,target))
```
