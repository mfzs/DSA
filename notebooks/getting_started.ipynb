# Sum of Two array from the list with a target

```python

def sum(list,target):
    n=len(list)
    for i in range(n-1):
        for j in range(i+1,n):
            if list[i]+list[j]==target:
                return [i,j]
    return "Not found"

print(sum([2,7,11,15],22))
```

```python
def sum(list,target):
    n=len(list)
    tempMap={}
    # temp=0
    for i in range(n):
        tempMap[list[i]]=i
    # print(tempMap)
    for i in range(n):
        temp=target-list[i]
        # print(temp)
        if temp in tempMap and tempMap[temp]!=i: # comment the code from and then it can reuse index as well
            return [i,tempMap[temp]]
    return "Not found"
 
    
print(sum([2,7,11,15,19],21))
```

```python
def sum(list,target):
    n=len(list)
    if n<=1:
        return "List is empty"
    for i in range(n):
        temp=target-list[i]
        # print(temp)
        try:
            tempIndex=list.index(temp)
            # print(tempIndex)
            if tempIndex!=i and 0 <= tempIndex < len(list) : 
                return [i,tempIndex]
        except ValueError:
            # print("Not found",i)
            continue

    return "Not found"
print(sum([1,22,0,0,22],22))
```

Anagrams

```python
def group_anagrams(strs):
    groups = {}
    
    for s in strs:
        # Sort string to get canonical form
        sorted_s = ''.join(sorted(s))
        # print(sorted_s)
        if sorted_s in groups:
            groups[sorted_s].append(s)
            # print(groups)
        else:
            groups[sorted_s] = [s]
            # print(groups)
    
    return list(groups.values())

# Test
strs = ["eat","tea","tan","ate","nat","bat"]
print(group_anagrams(strs))
```

```python
def anagram(strs):
    tempMap = {}
    for s in strs:
        sorted_value = ''.join(sorted(s))
        if sorted_value in tempMap:
            tempMap[sorted_value].append(s)
        else:
            tempMap[sorted_value] = [s]
    
    # Return groups with counts
    # print(len(tempMap))
    return list([(group, len(group)) for group in tempMap.values()])

print(anagram(["eat", "tea", "tan", "ate", "nat", "bat"]))

```

```python
def first_fit_decreasing(items, bin_capacity):
# Sort items in decreasing order
    items = sorted(items, reverse=True)
    # print(items) # output: [0.8, 0.7, 0.6, 0.5, 0.4, 0.3, 0.2]
    bins = []
    for item in items:
        placed = False
        # print(item)
        # Try placing the item in an existing bin
        for i in range(len(bins)):
            # print(bins[i])
            if bins[i] + item <= bin_capacity:
                bins[i] += item
                # print(bins,item)
                placed = True
                break

        # If not placed, create a new bin
        if not placed:
            # print("New bin",item)
            bins.append(item)

    return bins

items = [0.5, 0.7, 0.3, 0.5, 0.6, 0.2, 0.8,0.4,0.1]
bin_capacity = 1.0

bins = first_fit_decreasing(items, bin_capacity)

print(f"Number of bins used: {len(bins)}")
print("Bins filled with total used space:", bins)
```

```python

def reorderLogFiles(logs):
    # """
    # :type logs: List[str]
    # :rtype: List[str]
    # """
    digits = []
    letters = []
	# divide logs into two parts, one is digit logs, the other is letter logs
    for log in logs:
        if log.split()[1].isdigit():
            digits.append(log)
        else:
            letters.append(log)
            # print(log.split()[0])
                
    letters.sort(key=lambda log: log.split()[0])     # when suffix is tie, sort by identifier
    print(letters)
    letters.sort(key=lambda log: log.split()[1:])    # sorted by suffix   
    print(letters)
    result = letters + digits                                        #put digit logs after letter logs
    return result

print(reorderLogFiles(["dig1 8 1 5 1","let1 art can", "dig2 3 6","let2 own kit dig","let3 art zero"]))
```

```python
# Merge two sorted arrays

def sorted_array(arr1,arr2):
    # initialise two pointers
    i,j=0,0

    # initilise emptyarray 
    merged_array=[]

    while i<len(arr1) and j<len(arr2):
        if arr1[i]<=arr2[j]:
            merged_array.append(arr1[i])
            i+=1
        else:
            merged_array.append(arr2[j])
            j+=1    
    
    merged_array.extend(arr1[i:])
    merged_array.extend(arr2[j:])

    return merged_array

def sort_array(arr1,arr2):
    c=arr1+arr2
    c.sort()
    return c
a=[1,2,3,4,5,10,12,13]
b=[6,7,8,9,10,17,29]
c=[6,0,8,9,10,17,29]
print(sorted_array(a,b))
print(sort_array(a,c))

```
