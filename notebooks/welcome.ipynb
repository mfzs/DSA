```python
## Nested Loop
i,j=0,0
count=0
for i in range(4):
    for j in range(3):
        # print(i,j)
        sum=i+j
        print(sum)
        count=count+1
print(count)
```

```python
## Prime Numbers
num = int(input("Enter the number: "))

if num <= 1:
    print("Not prime")
else:
    is_prime = True

    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

    if is_prime:
        print("Prime")
    else:
        print("Not prime")
 
        
```

```python
def primeCheck(num):
    if num <2:
        return "Not Prime"
    if num ==2:
        return "Prime"
    for i in range (2,num):
        if num%i==0:
            return "Not Prime"
            break
    return "Prime"
            
primeCheck(18)
        
```
