```python
# Comparison Operator
## ==, !=, < , > 

a=10
b=10
if a==b:
    print("True")
else:
    print("False")

if a!=b:
    print("a is not equal to b")
else:
    print("A is equal to B")

if a>b:
    print("a is greater than b")
else:
    print("A is not greater than B")

if a<b:
    print("B is greater than A")
else:
    print("B is not greater than A")

```
