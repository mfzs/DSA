```python
## Modeling a bank Account

class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner=owner
        self.balance=balance

    def deposit(self,amount):
        self.balance+=amount
        return f"{amount} is deposited into the account of {self.owner},New balance is {self.balance}"

    def withdraw(self,amount):
        if amount>self.balance:
            return f"Not enough availble fund, Current Balance is {self.balance}"
        else:
            self.balance-=amount
            return f"{amount} is withdrawl from the account of {self.owner},New balance is {self.balance}"

    def getBalanace(self):
        return self.balance

firoz=BankAccount("Firoz")
# print(firoz)
Ruhee=BankAccount("Ruhee")
firoz.deposit(5000)
firoz.deposit(15000)
firoz.withdraw(3000)
print(firoz.getBalanace())

Ruhee.deposit(15000)
Ruhee.deposit(25000)
print(Ruhee.withdraw(7000))
print(Ruhee.getBalanace())
```

```python
## Inheritence

class Car():
    def __init__(self,windows,doors,enginetype):
        self.windows=windows
        self.doors=doors
        self.enginetype=enginetype

    def drive(self,person):
        return f"{person} would be driving {self.enginetype}"

firoz=Car(4,5,"Petrol")
print(firoz.drive("Ruhee"))


class Tesla(Car): # Single Inheritance
    def __init__(self,windows,doors,enginetype,is_selfdriving):
        super().__init__(windows,doors,enginetype) # basically calling init method of parent class
        self.is_selfdriving=is_selfdriving

    def selfdrive(self):
        return f"Support self Drive?: {self.is_selfdriving}"

Ruhee=Tesla(2,2,"EV",True)
print(Ruhee.selfdrive())
print(Ruhee.drive("Ruhee"))


```

```python
## Multiple Inheritance - Having more than one base class

class Animal:
    def __init__(self,name):
        self.name=name
    def speak(self):
        return "Subclass need to have this method"

class Pet:
    def __init__(self,owner):
        self.owner=owner

class Dog(Animal,Pet):
    def __init__(self,name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)
    def speak(self):
        return f"{self.name} say woof"

dog1=Dog("Buddy","Firoz")
print(dog1.speak())
print(dog1.owner)
```

# Polymorphism
- Multiple forms
- Achieve using method overriding and interfaces
- Objects of diff classes to be treated as object of common class

```python
class Shape:
    def area(self):
        return "Area of Shape:"

class Rectangle:
    def __init__(self,width,breadth):
        self.width=width
        self.breadth=breadth

    def area(self):
        return f"Area would be {self.width * self.breadth}"

class Circle(Shape):
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        return f"area would be {self.radius*3.14}"


def print_area(shape):
    return f"{shape.area()}"

rectangle1=Rectangle(4,5)
# print_area(rectangle1)
print(print_area(rectangle1))

circle=Circle(5)
print(print_area(circle))
```

```python
## Abstract base classes (interfaces in other languages)

from abc import ABC,abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

class Car:
    def start_engine(self):
        return "Car started"

class Bike:
    def start_engine(self):
        return "Bike started"

def start_engine(vehicle):
        return vehicle.start_engine()

bike=Bike()
print(bike.start_engine())

car=Car()
print(car.start_engine())
```

# Operator Overloading
- Changing the default functionality of a opertaor

```python
class Vector:
    def __init__(self,x,y):
        self.x=x
        self.y=y

    def __add__(self,other):
        return Vector(self.x+other.x,self.y+other.y)
    def __sub__(self,other):
        return Vector(self.x-other.x,self.y-other.y)
    def __mul__(self,other):
        return Vector(self.x*other.x,self.y*other.y)
    def __eq__(self,other):
        return self.x== other.x and self.y==other.y
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

v1=Vector(2,3)
v2=Vector(4,5)
print(v1+v2)
print(v1-v2)
print(v1*v2)
print(v1==v2)
```

```python
''' Generators
- Generators are simple way for creating iterators
- 'yield' keyword to produce values lazily and do not store them in memory
-  e.g.  
'''

def square(n):
    for i in range(3):
        return i**2
print(square(3))

def square(n):
    for i in range(3):
        yield i**2
print(square(3))

for i in square(3):
    print(i)
a=square(3)
next(a)
```

```python
## Reading large files

def read_large_file(file_path):
    with open(file_path,'r') as file:
        for line in file:
            yield line

file_path='sample.txt'
for line in read_large_file(file_path):
    print(line.strip())
```

Decorators
- To modify the behaviour of a function and class
- Add additonal functionality without changing/modifying the actual function
- starting using '@'

```python
def welcome(func):
    def sub_welcome():
        print("Statement 1")
        func()
        print("Statement 2")
        print("---------------")
    return sub_welcome()
welcome(print)

def dec_func():
    print("New function to test decorator")
welcome(dec_func)
```

```python
def welcome(func):
    def sub_welcome():
        print("Statement 1")
        func()
        print("Statement 2")
        print("---------------")
    return sub_welcome()
    
@welcome # dec_func will go as a param inside the welcome function
def dec_func():
    print("New function to test decorator")
```
