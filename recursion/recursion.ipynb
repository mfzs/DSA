# Recursion
- We can call a function inside the function, if a function calls itself it is recursion
- A function stays in the memory until it gets fully executed
- Why python has recursion overflow?
- - RecursionError: maximum recursion depth exceeded
  - Python enforces it to avoid crashing your system
  - maximum recursion depth (~1000 by default).
  - import sys
  - sys.setrecursionlimit(2000)
    

```python
def printNum(num):
    print(num)
    if num <= 1:
        return
    printNum(num-1)

printNum(5)
```

```python
def facto(num):
    if num<0:
        return "Negative number"
    if num==1:
        return 1
    if num==0:
        return 1
    return num*(facto(num-1))
facto(6)
```

# PMI - Principle of Mathematical Induction
- Recursion is based on PMI
- Used to prove some fact.
- - f(n)=True for all natural numbers
  - ## Three steps:
    1. Prove f(0) or f(1) is true
    2. Assume f(k) is true -> Induction Hypothesis
    3. Using step 2., prove f(k+1) is also true
  - 

```python
def powerof2(power):
    if power==0:
        return 1
    if power==1:
        return 2
    return 2*powerof2(power-1)
        
print(powerof2(4))
```

```python
def powerof(num,power):
    if num==1:
        return 1
    if power==1:
        return num
    return num*powerof(num,power-1)
        
print(powerof(4,3))
```

```python
## Sum of numbers
def sumf(num):
    if num==0:
        return 0
    return num+sumf(num-1)
sumf(5)
```

```python
## No. of Digits 
def digits(num):
    if num==0:
        return 0
    if num>=1 and  num<=9:
        return 1
    # smallans=int(digits(num/10))
    # ans=1+smallans
    return 1+int(digits(num/10))
digits(444)
```

```python
## Fibonacci Series
# functioncall=1
def fibonacci_series(num):
    # print(num)
    # global functioncall
    
    # print(functioncall, "for", num,"\n")
    # functioncall+=1
    
    if num==0 or num==1:
        return num
    return fibonacci_series(num-1)+fibonacci_series(num-2)

print(fibonacci_series(6))
for i in range(1,7):
    print(fibonacci_series(i), end=" ")
```

```python
counter=0
def fact_head(n):
    global counter
    if n==0:
        return 1
    counter+=1
    print(counter)
    smallans=fact_head(n-1)
    finalAns=n*smallans
    return finalAns
print(fact_head(5))
```

```python
counter=0
def fact_tail_wrong(n): # Intentionally wrong way applied
    global counter
    if n==0:
        return 1
    counter+=1
    print(counter)
    return n*fact_tail_wrong(n-1)
print(fact_tail_wrong(5))
```

```python
counter=0
def fact_tail(n,accumalator=1):
    global counter
    if n==0:
        return accumalator
    accumalator=accumalator*n
    counter+=1
    print(counter)
    return fact_tail(n-1,accumalator)
print(fact_tail_wrong(5))
```

```python
def fibonacci_series(num,accumalator=1):
    
    if num==0 or num==1:
        return accumalator
    accumalator=accumalator+fibonacci_series(num-2)
    return fibonacci_series(num-1,accumalator)

print(fibonacci_series(5))
for i in range(1,6):
    print(fibonacci_series(i), end=" ")
```

```python
result=[]

def count_down(n):
    global result
    if n<=0:
        return result 
    result.append(n)
    result=count_down(n-1)
    return result
count_down(10)
```

```python
result=[]

def count_up(n):
    global result
    if n<=0:
        return []
    
    result=count_up(n-1)
    result.append(n)
    return result
count_up(5)
```
