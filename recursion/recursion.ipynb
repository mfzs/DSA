# Recursion
- We can call a function inside the function, if a function calls itself it is recursion
- A function stays in the memory until it gets fully executed
- Why python has recursion overflow?
- - RecursionError: maximum recursion depth exceeded
  - Python enforces it to avoid crashing your system
  - maximum recursion depth (~1000 by default).
  - import sys
  - sys.setrecursionlimit(2000)
    

```python
def printNum(num):
    print(num)
    if num <= 1:
        return
    printNum(num-1)

printNum(5)
```

# PMI - Principle of Mathematical Induction
- Recursion is based on PMI
- Used to prove some fact.
- - f(n)=True for all natural numbers
  - ## Three steps:
    1. Prove f(0) or f(1) is true
    2. Assume f(k) is true -> Induction Hypothesis
    3. Using step 2., prove f(k+1) is also true
  - 

```python
def powerof2(power):
    if power==0:
        return 1
    if power==1:
        return 2
    return 2*powerof2(power-1)
        
print(powerof2(4))
```

```python
def powerof(num,power):
    if num==1:
        return 1
    if power==1:
        return num
    return num*powerof(num,power-1)
        
print(powerof(4,3))
```

```python
## Sum of numbers
def sumf(num):
    if num==0:
        return 0
    return num+sumf(num-1)
sumf(5)
```

```python
# to check if array is sorted
'''
- base case
- smaller problem solution
- final solution
'''
def check_sorted_array(arr):
    size=len(arr)
    # Base case
    if size==1 or size==0:
        return True
    
    if arr[0]<=arr[1]:
        return check_sorted_array(arr[1:])
    else:
        return False
arr=[1,1,3,4,5]
check_sorted_array(arr)
```

```python
# First & last Index of the element using recursion
def first_index(arr,target):
    if len(arr)==0:
        return -1
    if arr[0]==target:
        return 0

    smallans=first_index(arr[1:],target)
    if smallans==-1:
        return -1
    else:
        return smallans+1
     
    
arr=[2,4,5,6,8,8,2,9,11,5,8]
target=9

first_index(arr,target)
```

```python
def last_index(arr, target):
    if len(arr) == 0:
        return -1
    
    smallans = last_index(arr[1:], target)
    
    if smallans != -1:
        return smallans + 1
    
    if arr[0] == target:
        return 0
    
    return -1


arr = [2,4,5,6,8,8,2,9,11,5,8]
target = 8

print(last_index(arr, target))
```

```python
def all_index(arr, target):
    if len(arr) == 0:
        return []
    
    smallans = all_index(arr[1:], target)
    smallans = [i + 1 for i in smallans]
    
    if arr[0] == target:
        return [0] + smallans
    
    return smallans


arr = [2,4,5,6,8,8,2,9,11,5,8]
target = 8
print(all_index(arr, target))
```

```python
def all_index(arr, target, i=0):
    if i == len(arr):
        return []
    
    smallans = all_index(arr, target, i+1)
    
    if arr[i] == target:
        return [i] + smallans
    
    return smallans

arr = [2,4,5,6,8,8,2,9,11,5,8]
target = 8
print(all_index(arr, target))
```

```python
# Count occurance

def count_occ(arr,target,i=0):
    count=0
    # 1. Base Case
    if i == len(arr):
        return 0

    # 2. Smaller Answer
    smallAns=count_occ(arr,target,i+1)

    # 3. Final Answer
    if arr[i]==target:
        return smallAns+1
    else:
        return smallAns



arr = [2,4,5,6,8,8,2,9,11,5,8]
target = 8
print(count_occ(arr, target))
```

```python
# First occurance

def first(arr,target,i=0):
    count=0
    # 1. Base Case
    if i == len(arr):
        return 0

    # 2. Smaller Answer
    smallAns=first(arr,target,i+1)

    # 3. Final Answer
    if arr[i]==target:
        return i
    return smallAns



arr = [2,4,5,6,8,8,2,9,11,5,8]
target = 8
print(first(arr, target))
```

```python
# Last occurance

def last(arr,target,i=0):
    count=0
    # 1. Base Case
    if i == len(arr):
        return -1

    # 2. Smaller Answer
    smallAns=last(arr,target,i+1)

    # 3. Final Answer
    if smallAns!=-1:
        return smallAns
    
    if arr[i] == target:
        return i
    
    return -1



arr = [2,4,5,6,8,8,2,9,11,5,8]
target = 2
print(last(arr, target))
```
